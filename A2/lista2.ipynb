```python
class Sensor():
    def __init__(self, id, x, y, r):
        self.id = id
        self.coordinates = (x, y)
        self.r = r

sensors = [
    Sensor(0, 0.0, 0.0, 30.0),
    Sensor(1, 25.0, 0.0, 25.0),
    Sensor(2, 50.0, 0.0, 30.0),
    Sensor(3, 15.0, 20.0, 35.0),
    Sensor(4, 35.0, 25.0, 25.0),
    Sensor(5, 75.0, 10.0, 15.0),
    Sensor(6, 70.0, 20.0, 20.0),
    Sensor(7, 55.0, 25.0, 25.0)
            ]

s_i = 0
s_f = 5

s =  [0, 1, 2, 5, 6, 7, 4, 3, 0]
```

```python
def questao2(sensors, vi, vj):    

    def calc_distance(sensor1: Sensor, sensor2: Sensor):
        x1, y1 = sensor1.coordinates
        x2, y2 = sensor2.coordinates
        return pow((x2 - x1)**2 + (y2 - y1)**2, exp= 1/2)

    graph = [[] for _ in range(len(sensors))]

    for sensor1 in sensors:
        for sensor2 in sensors:
            if sensor1.id != sensor2.id:
                distance = calc_distance(sensor1, sensor2)
                if distance <= sensor1.r:
                    graph[sensor1.id].append((sensor2.id, distance))

    class MinHeap():
        def __init__(self):
            self.data = []
            self.pos = {}

        def __len__(self):
            return len(self.data)

        def __swap__(self, i, j):
            self.data[i], self.data[j] = self.data[j], self.data[i]
            self.pos[self.data[i][1]] = i
            self.pos[self.data[j][1]] = j

        def heapify_up(self, idx):
            while idx > 0:
                pai = (idx - 1)//2
                if self.data[pai] > self.data[idx]:
                    self.__swap__(pai, idx)
                    idx = pai
                else:
                    break

        def heapify_down(self, idx):
            while idx*2 + 1 < len(self.data):
                l = idx*2 + 1
                r = idx*2 + 2
                menor = l
                if r < len(self.data) and self.data[r] < self.data[menor]:
                    menor = r 
                if self.data[idx] > self.data[menor]:
                    self.__swap__(menor, idx)
                    idx = menor
                else:
                    break
                
        def insert_or_update(self, node):
            v = node[1]
            w = node[0]

            if v in self.pos:
                idx = self.pos[v]
                if w < self.data[idx][0]:
                    self.data[idx] = (w, v)
                    self.heapify_up(idx)

            else:
                self.data.append(node)
                idx = len(self.data) - 1
                self.pos[v] = idx
                self.heapify_up(idx)

        def pop(self):
            node = self.data[0]
            self.data[0] = self.data[-1]
            self.data.pop()
            del self.pos[node[1]]

            if self.data:
                self.pos[self.data[0][1]] = 0
                self.heapify_down(0) 

            return node

    def Dijkstra(graph, vi, vj):
        parents = [-1 for _ in range(len(graph))]
        parents[vi] = vi
        distances = [float('inf') for _ in range(len(graph))]
        distances[vi] = 0

        heap = MinHeap()
        heap.insert_or_update((0, vi))

        while heap:
            node = heap.pop()
            w, v = node
            if v == vj:
                return True, parents
            for v_next, w in graph[v]:
                new_distance = distances[v] + w
                if new_distance < distances[v_next]:
                    distances[v_next] = new_distance
                    parents[v_next] = v
                    heap.insert_or_update((distances[v_next], v_next))

        return False, parents

    def find_order(parents, vi, vj):
        order = []
        pai = vj
        while pai != vi:
            order.append(pai)
            pai = parents[pai]
        order.append(vi)
        return list(reversed(order))


    flag1, parents1 = Dijkstra(graph, s_i, s_f)
    flag2, parents2 = Dijkstra(graph, s_f, s_i)

    if flag1 and flag2:
        order1 = find_order(parents1, vi, vj)
        order2 = find_order(parents2, vj, vi)

        print(f'{order1 + order2[1:]}')

    else:
        print('Vai dar não!')

```
