```python
cruzamentos = [(1, 2, 1), (1, 4, 2), (2, 3, 7),
                (2, 5, 1), (3, 6, 9), (4, 5, 3),
                (4, 7, 2), (5, 6, 1), (5, 8, 2),
                (6, 9, 10), (7, 8, 1), (8, 9, 1)]

def question_1(cruzamentos, C):
    import heapq
    from collections import defaultdict

    graph = defaultdict(list)
    for cruzamento_1, cruzamento_2, custo in cruzamentos:
        graph[cruzamento_1].append((cruzamento_2, custo))
        graph[cruzamento_2].append((cruzamento_1, custo))

    inicio = list(graph)[0]
    visitados = {inicio}
    heap = [(w, v) for v, w in graph[inicio]]
    heapq.heapify(heap)
    total_cost = 0
    
    while heap:
        w, v = heapq.heappop(heap)

        if v in visitados:
            continue
        visitados.add(v)
        
        total_cost += w
        if len(visitados) == len(graph):
            return total_cost <= C

        for v_next, w in graph[v]:
            if v_next in visitados:
                continue

            heapq.heappush(heap, (w, v_next))

    return False

print(question_1(cruzamentos, 21))
```

```python
cruzamentos = [(1, 2, 1), (1, 4, 2), (2, 3, 7),
                (2, 5, 1), (3, 6, 9), (4, 5, 3),
                (4, 7, 2), (5, 6, 1), (5, 8, 2),
                (6, 9, 10), (7, 8, 1), (8, 9, 1)]

def question_1(cruzamentos, C):
    from collections import defaultdict

    graph = defaultdict(list)
    for cruzamento_1, cruzamento_2, custo in cruzamentos:
        graph[cruzamento_1].append((cruzamento_2, custo))
        graph[cruzamento_2].append((cruzamento_1, custo))

    V = len(graph)
    parents = [v for v in range(V)]
    sets = [1 for _ in range(V)]

    def find(v):
        pai = parents[v]
        while pai != parents[pai]:
            pai = parents[pai]
        return pai

    def union(v1, v2):
        parent1 = find(v1)
        parent2 = find(v2)

        if sets[parent1] > sets[parent2]:
            sets[parent1] += sets[parent2]
            parents[parent2] = parent1
        else:
            sets[parent2] += sets[parent1]
            parents[parent1] = parent2

    edges_list = sorted([(w, cruzamento_1 - 1, cruzamento_2 - 1) for cruzamento_1, cruzamento_2, w in cruzamentos])
    total_cost = 0
    inTree_edges = 0

    for w, v1, v2 in edges_list:
        if find(v1) != find(v2):
            union(v1, v2)
            total_cost += w
            inTree_edges += 1
        if inTree_edges == V - 1:
            return total_cost <= C

    return False

print(question_1(cruzamentos, 21))
```
