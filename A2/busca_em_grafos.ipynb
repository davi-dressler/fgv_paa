```python
def dfs(graph, s):
    stack = [(s, False)]

    V = len(graph)
    pre_order = [-1]*V
    count_pre = 0
    pos_order = [-1]*V
    count_pos = 0

    while stack:
        v, finished = stack.pop()

        if finished:
            for v_next in graph[v]:
                if pos_order[v_next] == -1:
                    print("O grafo tem um ciclo!")
                    return
            pos_order[v] = count_pos
            count_pos += 1
            continue

        if pre_order[v] != -1:
            continue
    
        pre_order[v] = count_pre
        count_pre += 1

        stack.append((v, True))

        for v_next in graph[v]:
            if pre_order[v_next] == -1:
                stack.append((v_next, False))

    

dfs(graph, 0)

```

```python
def is_Bipartite(graph: list[list[int]]):
    set = [-1 for _ in range(len(graph))]

    for s in range(len(graph)):
        if set[s] != -1:
            continue
        head = 0
        queue = [s]
        set[s] = 0

        while head < len(queue):
            v = queue[head]
            head += 1

            for v_next in graph[v]:
                if set[v_next] == -1:
                    set[v_next] = abs(set[v] - 1)
                    queue.append(v_next)
                elif set[v] == set[v_next]:
                    return False

        return True

is_Bipartite(graph)


```

```python
class MinHeap:
    def __init__(self):
        self.data = []

    def insert(self, node):
        self.data.append(node)
        idx = len(self.data) - 1

        while idx > 0:
            pai = (idx - 1)//2
            if self.data[pai] > self.data[idx]:
                self.data[pai], self.data[idx] = self.data[idx], self.data[pai]
                idx = pai
            else:
                break

    def heapify_down(self, idx):
        while idx*2 + 1 < len(self.data):
            l = idx*2 + 1
            r = idx*2 + 2
            menor = l

            if r < len(self.data) and self.data[r] < self.data[l]:
                menor = r

            if self.data[idx] > self.data[menor]:
                self.data[menor], self.data[idx] = self.data[idx], self.data[menor]
                idx = menor

            else:
                break

    def pop(self):
        node = self.data[0]
        self.data[0] = self.data[-1]
        self.data.pop()
        self.heapify_down(0)
        return node

    
```

```python
class MinHeap():

    def __init__(self):
        self.data = []
        self.pos = {}

    def __len__(self):
        return len(self.data)
    
    def __contains__(self, node):
        v = node[1]
        return v in self.pos

    def _swap(self, i, j):
        self.data[i], self.data[j] = self.data[j], self.data[i]
        self.pos[self.data[i][1]] = i
        self.pos[self.data[j][1]] = j

    def _heapify_up(self, idx):
        while idx > 0:
            pai = (idx - 1)//2
            if self.data[pai] > self.data[idx]:
                self._swap(pai, idx)
                idx = pai
            else:
                break

    def _heapify_down(self, idx):
        while idx*2 + 1 < len(self.data):
            l = idx*2 + 1
            r = idx*2 + 2
            menor = l
            if r < len(self.data) and self.data[r] < self.data[l]:
                menor = r
            if self.data[menor] < self.data[idx]:
                self._swap(menor, idx)
                idx = menor
            else:
                break

    def insert_or_update(self, node):
        v = node[1]
        w = node[0]

        # Update
        if v in self.pos: 
            idx = self.pos[v]
            if w < self.data[idx][0]:
                self.data[idx] = (w, v)
                self._heapify_up(idx)

        # Insert
        else:
            self.data.append(node)
            idx = len(self.data) - 1
            self.pos[v] = idx
            self._heapify_up(idx)

    def pop(self):
        node = self.data[0]
        self.data[0] = self.data[-1]
        self.data.pop()
        del self.pos[node[1]]

        if self.data:
            self.pos[self.data[0][1]] = 0
            self._heapify_down(0)

        return node
```

```python
def Djikstra(graph: list[list[int]], s: int):
    distances = [float('inf') for _ in range(len(graph))]
    distances[s] = 0
    visited = set()

    heap = MinHeap()
    heap.insert_or_update((0,s))

    while heap.data:
        d, v = heap.pop()
        if v in visited:
            continue
        visited.add(v)
        print("____________________________")
        print(f"Vértice processado: {v}")

        for v_next, w in graph[v]:
            print(f"Vértice visitado: {v_next}")
            dist_acum = d + w
            print(f"Distância acumulada: {distances[v_next]} | Distância analisada: {dist_acum}")
            if dist_acum < distances[v_next]: 
                distances[v_next] = dist_acum
                heap.insert_or_update((dist_acum, v_next))
                print(f"Heap = {heap.data}")
                print("Distância atualizada.")

        print(visited)
        print(distances)
        print("")

Djikstra(graph, 0)
```

# Algoritmo de Bellman-Ford

Esse algoritmo também encontra a árvore de menor caminho de um grafo a partir de uma fonte, mas a diferença entre este algoritmo e o de Djikstra é que este consegue lidar com pesos negativos.

```python
def Bellman_Ford(graph: list[list[int]], s):
    distances = [float('inf') for _ in range(len(graph))]
    distances[s] = 0

    len_path = [0 for _ in range(len(graph))]

    queue = [s]
    head = 0

    in_queue = [0 for _ in range(len(graph))]
    in_queue[s] = 1
    
    while head < len(queue):
        v = queue[head]
        head += 1
        in_queue[v] = 0

        for v_next, w in graph[v]:
            print(f"Aresta analisada: ({v, v_next})")
            if distances[v] + w < distances[v_next]:
                distances[v_next] = distances[v] + w
                len_path[v_next] = len_path[v] + 1
                print(f"Aresta relaxada: {distances}")

                if len_path[v_next] >= len(graph):
                    return False

                if in_queue[v_next] != 1:
                    queue.append(v_next)
                    in_queue[v_next] = 1
                

    return distances
```

# Algoritmo de Prim

Esse algoritmo é usada para encontrar a árvore geradora mínima de um grafo G a partir de um vértice fonte

```python
def Prim(graph: list[list[int]], s):
    n = len(graph) 
    inTree = [False for _ in range(n)]
    vertexCost = [float('inf') for _ in range(n)]
    parents = [-1 for _ in range(n)]

    # Inicialização
    vertexCost[s] = 0
    parents[s] = s

    heap = MinHeap()
    heap.insert_or_update((0, s))

    while heap:
        _, v = heap.pop()
        inTree[v] = True

        for v_next, w in graph[v]:
            if not inTree[v_next] and w < vertexCost[v_next]:
                heap.insert_or_update((w, v_next))
                vertexCost[v_next] = w 
                parents[v_next] = v
                print(f"{parents}")

    total_cost = sum(vertexCost[v] for v in range(n) if inTree[v])
    print(f"{parents} \nCusto total = {total_cost}")
    
    return parents, total_cost
```

# Algoritmo de Kruskal

O objetivo do algoritmo de Kruskal é conseguir encontrar a árvore geradora mínima de um grafo analisando as arestas do grafo, pegamos a aresta com menor peso e adicionamos ao subgrafo, continuamos fazendo isso até que tenhamos um subgrafo de G que seja uma árvore e que contenha todos os vértices de G. 

Antes de prosseguir para a implementação do algoritmo, precisamos trabalhar na implementação de uma estrutura de dados chamada Union-Find.

```python
class UnionFind():
    def __init__(self, num_elements):
        self.parent = list(range(num_elements))
        self.group_size = [1 for _ in range(num_elements)]

    def find(self, v):
        leader = v
        while leader != self.parent[leader]:
            leader = self.parent[leader]

        return leader

    def union(self, v1, v2):
        root1, root2 = self.find(v1), self.find(v2)
        
        if root1 == root2: return False

        if self.group_size[root1] < self.group_size[root2]:
            self.parent[root1] = self.parent[root2]
            self.group_size[root2] += self.group_size[root1]
        else:
            self.parent[root2] = self.parent[root1]
            self.group_size[root1] += self.group_size[root2]

        return True
        
```

```python
def Kruskal(graph: list[list[tuple[int, int]]]):
    n = len(graph)

    edges = sorted((w, v, v_next) for v in range(n) for v_next, w in graph[v])

    tree = UnionFind(num_elements= n)
    mst = []
    total_weight = 0

    for w, v, v_next in edges:
        if tree.union(v, v_next):
            mst.append((v, v_next, w))
            total_weight += w
            if len(mst) == n - 1:
                break

    return total_weight, mst
        
    

    
```
